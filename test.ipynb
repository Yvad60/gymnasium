```typescript
import { AzureChatOpenAI } from '@langchain/openai'
import 'dotenv/config'

const model = new AzureChatOpenAI({
  azureOpenAIApiKey: process.env.AZURE_OPENAI_API_KEY,
  azureOpenAIEndpoint: process.env.AZURE_OPENAI_ENDPOINT,
  azureOpenAIApiVersion: process.env.AZURE_OPENAI_API_VERSION,
  azureOpenAIApiDeploymentName: process.env.AZURE_OPENAI_DEPLOYMENT
})
```

```typescript
import { createAgent, HumanMessage, AIMessage } from 'langchain'
import {MemorySaver} from '@langchain/langgraph'


const agent = createAgent({
  model,
  systemPrompt: "You are an assistant that helps answer questions. You use french accent, with french jokes and slangs",
  checkpointer: new MemorySaver()
})

const event = await agent.invoke({
  messages: [new HumanMessage('Hello there'), new AIMessage('General Kenobi'), new HumanMessage('Who is that')],
}, {
  configurable: {"thread_id": 'test-thread'}
})

console.log(event.messages)

```

```typescript
import {  HumanMessage } from 'langchain'

const response = await agent.invoke({
  messages: [new HumanMessage('Who are we talking about again?')],
},{
  configurable: {"thread_id": 'test-thread'}
})
console.log(response.messages)
```
